```python
import cv2
import numpy as np

# Load two sample images
image1 = cv2.imread("ganderbal3.jpg")
image2 = cv2.imread("ganderbal2.jpg")

# Check whether images are loaded successfully
if image1 is None or image2 is None:
    print("Error: Unable to load one or both images.")
else:
    # Get dimensions of both images
    height1, width1 = image1.shape[:2]
    height2, width2 = image2.shape[:2]

    print("Image 1 Dimensions:", width1, "x", height1)
    print("Image 2 Dimensions:", width2, "x", height2)

    # Resize image2 if dimensions do not match
    if image1.shape[:2] != image2.shape[:2]:
        image2 = cv2.resize(image2, (width1, height1))
        print("Image 2 resized successfully.")

    # Display both images
    cv2.imshow("Image 1", image1)
    cv2.imshow("Image 2", image2)

     # Perform weighted image blending
    # alpha = 0.6, beta = 0.4, gamma = 0
    blended_image = cv2.addWeighted(
        image1, 0.6,
        image2, 0.4,
        0
    )
    
    # Display the blended image
    cv2.imshow("Blended Composite Image", blended_image)

    # Save the output image
    cv2.imwrite("blended_image.jpg", blended_image)


     # Combine images horizontally for verification
    combined = np.hstack((image1, image2, blended_image))

    # Display all three images
    cv2.imshow("Original 1 | Original 2 | Blended Output", combined)

    # Save comparison image
    cv2.imwrite("image_comparison.jpg", combined)


    cv2.waitKey(0)
    cv2.destroyAllWindows()
    

    print("Blended image generated and saved successfully.")

    


 #   Explanation
#1. cv2.imread() loads the two image files.
#2. image.shape[:2] retrieves image height and width.
#3. cv2.resize() adjusts the second image to match the first image's dimensions.
#4. cv2.imshow() displays both images.
#5. cv2.waitKey(0) waits for a key press.
#6. cv2.destroyAllWindows() closes all display windows.
```
