```python

import gradio as gr

# Define chatbot response function
def chatbot_response(message, history):

    message = message.lower().strip()

    if "hello" in message or "hi" in message:
        return "Hello! Welcome to my Python Gradio Chatbot. How can I help you?"

    elif "python" in message:
        return "Python is a popular programming language used in AI, Data Science, Web Development and Automation."

    elif "gradio" in message:
        return "Gradio is a Python library used to create interactive web interfaces for machine learning applications."

    elif "name" in message:
        return "I am an interactive chatbot developed using Python and Gradio."

    elif "thank" in message:
        return "You are welcome! Feel free to ask more questions."

    else:
        return "Thank you for your question. Please ask me about Python or Gradio."

# Create chatbot interface
demo = gr.ChatInterface(
    fn=chatbot_response,
    title="Python Interactive Chatbot",
    description="""
    Welcome to the Python Gradio Chatbot!

    You can interact with this chatbot by typing a message
    or selecting one of the sample prompts below.
    """,
    examples=[
        "Hello! Introduce yourself.",
        "What is Python used for?",
        "What is Gradio?"
    ],
    chatbot=gr.Chatbot(height=400),
    textbox=gr.Textbox(
        placeholder="Type your message here...",
        container=True
    )
)

# Launch chatbot
if __name__ == "__main__":
    demo.launch()
```
