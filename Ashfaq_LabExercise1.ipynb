```python
import pandas as pd
import numpy as np

# Generate random numerical data
data = np.random.randint(1, 101, size=(20, 4))

# Create DataFrame
df = pd.DataFrame(data, columns=['f1', 'f2', 'f3', 'f4'])

# Display DataFrame
print(df)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.scatter(df['f2'], df['f3'])

plt.xlabel('Column f2')
plt.ylabel('Column f3')

plt.title('Scatter Plot: Relationship between f2 and f3')

plt.grid(True)
plt.show()
```

```python
# Calculate row-wise arithmetic mean
df['f5'] = df[['f1', 'f2', 'f3', 'f4']].mean(axis=1)

# Display updated DataFrame
print(df)
```

```python
df.to_excel("dataframe_f1_to_f5.xlsx", index=False)
```
